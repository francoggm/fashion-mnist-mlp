```python
import numpy as np
import matplotlib.pyplot as plt

def load_image(file):
    with open(file, 'rb') as f:
        f.seek(16)  # Skip the header
        data = np.frombuffer(f.read(), dtype=np.uint8)
    return data.reshape(-1, 28, 28)  # Reshape to (num_images, height, width)

def load_label(file):
    with open(file, 'rb') as f:
        f.seek(8)  # Skip the header
        data = np.frombuffer(f.read(), dtype=np.uint8)
    return data

path = '../assets/raw'
X_train = load_image(f'{path}/train-images-idx3-ubyte')
Y_train = load_label(f'{path}/train-labels-idx1-ubyte')
```

```python
CLASSES = ["T-shirt/top","Trouser","Pullover","Dress","Coat",
              "Sandal","Shirt","Sneaker","Bag","Ankle boot"]

fig, axes = plt.subplots(3, 5, figsize=(10, 6))
for ax, img, label in zip(axes.flat, X_train[:15], Y_train[:15]):
    ax.imshow(img, cmap="gray")
    ax.set_title(CLASSES[label], fontsize=9)
    ax.axis("off")
plt.tight_layout()
plt.show()
```
